```python
# CELL 1 — Create Catalog
# Run this once. If it already exists, it will be skipped (IF NOT EXISTS is idempotent)
spark.sql("""
    CREATE CATALOG IF NOT EXISTS restaurant_catalog
    COMMENT 'Restaurant Analytics - Indian Chain UAE | Databricks Free Edition Project'
""")
print("✅ Catalog: restaurant_catalog")
```

```python
# CELL 2 — Create Medallion Schemas
schemas = {
    "landing": "Raw data landing zone — Unity Catalog Volume for CSV files + streaming Delta table",
    "bronze":  "Bronze layer — raw ingested data, DLT Auto Loader output, no business logic",
    "silver":  "Silver layer — cleaned, typed, validated star schema (fact + dim tables)",
    "gold":    "Gold layer — pre-aggregated tables for dashboard consumption",
}

for schema, comment in schemas.items():
    spark.sql(f"""
        CREATE SCHEMA IF NOT EXISTS restaurant_catalog.{schema}
        COMMENT '{comment}'
    """)
    print(f"✅ Schema: restaurant_catalog.{schema}")
```

```python
# CELL 3 — Create Unity Catalog Volume (replaces ADLS Gen2 mount point)
# Production standard: UC Volumes are the recommended way to manage files in Databricks
# They provide governance, lineage, and access control without mount points
spark.sql("""
    CREATE VOLUME IF NOT EXISTS restaurant_catalog.landing.raw_files
    COMMENT 'Raw CSV data files — customers, restaurants, menu items, historical orders, reviews'
""")
print("✅ Volume: /Volumes/restaurant_catalog/landing/raw_files/")
```

```python
# CELL 4 — Validate Everything
print("=== CATALOG VALIDATION ===\n")
display(spark.sql("SHOW SCHEMAS IN restaurant_catalog"))
display(spark.sql("SHOW VOLUMES IN restaurant_catalog.landing"))
print("✅ Validation complete. All schemas and volume ready.")
```
